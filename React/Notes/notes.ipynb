# Introduction

### Introduction to React

- **React** is a JavaScript **_library_** for building user interfaces, particularly single-page applications where you need a fast and interactive user experience. _( A single-page application is a web application that loads a single HTML page and dynamically updates the content as the user interacts with the app. In simple terms, it allows you to create a web application that feels more like a desktop application, with smooth transitions and without the need to reload the entire page for every interaction. )_

- **Library** vs **Framework**: A **_library_** is a collection of pre-written code that you can use to perform common tasks, while a **_framework_** provides a more structured way to build applications and often dictates the architecture of your application. React is considered a library because it focuses on the view layer of an application and allows developers to choose other libraries for handling different aspects of the application, such as state management or routing.

- In simpler terms, a **_library_** is like a toolbox that you can use to build something, while a **_framework_** is like a blueprint that guides you on how to build something. React provides the tools to build user interfaces, but it doesn't dictate how you should structure your application or handle other aspects of development.

- Reasons for using React:
  - **Component-Based Architecture**: React allows you to build encapsulated components that manage their own state and can be composed to create complex user interfaces. This promotes reusability and makes it easier to manage and maintain your code.
  - **Virtual DOM**: React uses a virtual DOM to optimize updates and rendering. When the state of a component changes, React updates the virtual DOM first, compares it with the previous version, and then efficiently updates only the parts of the actual DOM that have changed. This results in faster performance and a smoother user experience.
  - **Declarative Syntax**: React uses a declarative syntax, which means you describe what you want to see on the screen, and React takes care of updating the UI to match that description. This makes it easier to understand and reason about your code, as you can focus on the desired outcome rather than the steps needed to achieve it.
  - **Strong Community and Ecosystem**: React has a large and active community, which means there are plenty of resources, libraries, and tools available for developers. This makes it easier to find solutions to common problems and stay up-to-date with best practices in React development.
  - **Cross-Platform Development**: With React Native, you can use the same React components and concepts to build mobile applications for both iOS and Android platforms. This allows for code reuse and a consistent development experience across different platforms.
  - **SEO-Friendly**: React can be rendered on the server using frameworks like Next.js, which allows for better search engine optimization (SEO) compared to traditional client-side rendering. This is because search engines can easily crawl and index server-rendered content.
  - **It's not opiniated**: React is unopinionated, meaning it doesn't enforce a specific way of structuring your application or handling state management. This gives developers the freedom to choose the tools and libraries that best fit their needs and preferences, allowing for greater flexibility in how you build your applications.

### Setting Up a React Environment

##### Intro

- React can be used in projects in three ways:
    1. **Using a CDN**: You can include React in your HTML file using a Content Delivery Network (CDN) link. This is the simplest way to get started with React, but it is not recommended for production applications.Example:
    ```html
        <script src="https://unpkg.com/react@17/umd/react.development.js"></script>
        <script src="https://unpkg.com/react-dom@17/umd/react-dom.development.js"></script>
    ```

    2. **Using Toolchains**: You can use a toolchain like Create React App, Vite, or Next.js to set up a React project with a build system, development server, and other features. This is the recommended way to create React applications for production. Example:
    ```bash
        npx create-react-app my-app
        cd my-app
        npm start
    ```

    3. **Using Frameworks**: You can use a framework like Next.js or Gatsby to build React applications with additional features like server-side rendering, static site generation, and routing. This is a good option if you need more advanced features for your application. Example:
    ```bash
        npx create-next-app my-app
        cd my-app
        npm run dev
    ```

- **Note**: The choice of method depends on the complexity of your project and your specific requirements. For simple projects or learning purposes, using a CDN is sufficient. For larger projects or production applications, using a toolchain or framework is recommended.
    1. **Vite**: Vite is a modern build tool that provides a fast development experience for React applications. It uses native ES modules and supports hot module replacement (HMR) for faster updates during development. To create a new React project with Vite, you can use the following command:
    ```bash
        npm create vite@latest my-app --template react
        cd my-app
        npm install
        npm run dev
    ```

    2. **Next.js**: Next.js is a React framework that provides server-side rendering, static site generation, and other features for building production-ready applications. It is a good choice if you need advanced features like routing, API routes, and server-side rendering. To create a new Next.js project, you can use the following command:
    ```bash
        npx create-next-app my-app
        cd my-app
        npm run dev
    ```

    3. **Gatsby**: Gatsby is a React-based framework for building static websites and web applications. It provides a rich ecosystem of plugins and themes for building fast and optimized websites. To create a new Gatsby project, you can use the following command:
    ```bash
        npx gatsby new my-app
        cd my-app
        npm run develop
    ```

##### Delving Deeper into the React Setup

- Once you setup a React environment using vite, here are the folders that will be available:
    1. **node_modules**: This folder contains all the dependencies and packages that your project relies on. It is automatically generated when you run `npm install` and should not be modified manually.

    2. **public**: This folder contains static assets that are publicly accessible. It typically includes the `index.html` file, which is the entry point of your React application, as well as other static files like images, fonts, icons, and information files. The contents of this folder are served as-is and can be accessed directly via the URL.

    3. **src**: This folder contains the source code of your React application. It is where you will write your React components, styles, and other JavaScript files. The `src` folder is the main working directory for your application, and it is where you will spend most of your time developing your React app.
        1. **App.jsx**: This file is the main component of your React application. It serves as the root component that renders other components and manages the overall structure of your app. In simple terms, it is the starting point of your React application where you can define the layout and structure of your app.

        2. **main.jsx**: This file is the entry point of your React application. It is responsible for rendering the `App` component into the DOM. In simple terms, it is the file that connects your React application to the HTML file and allows it to be displayed in the browser.

        3. **index.css**: This file contains the global styles for your React application. It is where you can define the overall look and feel of your app, including colors, fonts, and layout styles. In simple terms, it is the file where you can customize the appearance of your React application.

        4. **assets**: This folder is typically used to store static assets such as images, fonts, and icons that are used in your React application. It is a common convention to keep all your static files organized in this folder for easy access and management.

        5. **App.css**: This file contains the styles specific to the `App` component. It is where you can define the styles that are applied only to the `App` component and its child components. In simple terms, it is the file where you can customize the appearance of the `App` component in your React application.
    
    4. **package.json**: This file contains metadata about your project, including the project name, version, dependencies, scripts, and other configuration settings. It is used by npm to manage your project's dependencies and scripts. In simple terms, it is the file that defines the structure and configuration of your React application.

    5. **package-lock.json**: This file is automatically generated by npm and contains a detailed description of the exact versions of dependencies installed in your project. It ensures that the same versions of dependencies are installed when the project is set up on different machines or environments. In simple terms, it is the file that locks the versions of your project's dependencies to ensure consistency across different setups.

    6. **vite.config.js**: This file contains the configuration settings for Vite, the build tool used in your React project. It allows you to customize various aspects of the build process, such as plugins, server settings, and build options. In simple terms, it is the file where you can configure how your React application is built and served during development and production.

    7. **.gitignore**: This file specifies which files and directories should be ignored by Git version control. It is used to prevent certain files, such as build artifacts, temporary files, and sensitive information, from being tracked in the Git repository. In simple terms, it is the file that tells Git which files to ignore when committing changes to your React application.

    8. **README.md**: This file contains documentation and information about your project. It is typically used to provide an overview of the project, installation instructions, usage guidelines, and other relevant information for developers and users. In simple terms, it is the file that serves as a guide and reference for understanding and using your React application.

    9. **eslint.config.js**: This file contains the configuration settings for ESLint, a tool used to analyze and enforce coding standards in your JavaScript code. It allows you to define rules and guidelines for writing clean and consistent code. In simple terms, it is the file where you can configure how your JavaScript code is checked for errors and adherence to coding standards in your React application.

# Getting Started with React

### React Components

- **Components** are the building blocks of a React application. They are reusable pieces of code that define how a part of the user interface should look and behave. Components can be thought of as custom HTML elements that encapsulate their own structure, style, and behavior.

- Components can be created as either **class components** or **functional components**. Class components are defined using ES6 classes and have access to lifecycle methods, while functional components are defined as JavaScript functions and can use hooks for managing state and side effects.

- Components syntax:
    1. **Class Component**:
    ```javascript
        import React, { Component } from 'react';

        class MyComponent extends Component {
            render() {
                return (
                    <div>
                        <h1>Hello, World!</h1>
                    </div>
                );
            }
        }

        export default MyComponent;
    ```

    2. **Functional Component**:
    ```javascript
        import React from 'react';

        const MyComponent = () => {
            return (
                <div>
                    <h1>Hello, World!</h1>
                </div>
            );
        }

        export default MyComponent;
    ```

- **Class vs Functional Components:**
    - **Class Components**:
        - Are defined using ES6 classes.
        - Have access to lifecycle methods (e.g., `componentDidMount`, `componentDidUpdate`).
        - Can manage state using `this.state` and `this.setState()`.
        - Tend to be more verbose and require more boilerplate code.
        - Uses:
            1. **State Management**: Class components can manage their own state using `this.state` and update it using `this.setState()`. This allows for dynamic rendering of the component based on changes in state.
            
            2. **Lifecycle Methods**: Class components have access to lifecycle methods, which allow developers to perform actions at specific points in a component's lifecycle, such as when it is mounted, updated, or unmounted. This can be useful for tasks like fetching data, setting up subscriptions, or cleaning up resources.

            3. **Complex Logic**: Class components can handle more complex logic and state management, making them suitable for larger applications or components with intricate behavior.


    - **Functional Components**:
        - Are defined as JavaScript functions.
        - Use hooks (e.g., `useState`, `useEffect`) to manage state and side effects.
        - Tend to be simpler and more concise.
        - Are the preferred way to create components in modern React development.
        - Uses:
            1. **Simplicity**: Functional components are simpler and more concise than class components, making them easier to read and understand. They are often preferred for smaller, stateless components that don't require complex logic or state management.

            2. **Hooks**: Functional components can use hooks, which allow developers to manage state and side effects in a more straightforward way. Hooks like `useState` and `useEffect` provide a cleaner and more intuitive way to handle state and lifecycle events compared to class components.

            3. **Performance**: Functional components can have better performance than class components in certain scenarios, as they don't have the overhead of class instances and can be optimized by React's rendering engine. This can lead to faster rendering times and improved performance in larger applications.

**NB:** React Components **_Must be Capitalized_**. This is because React _treats components that start with a lowercase letter as_ **_built-in HTML_** elements, while _components that start with an uppercase letter are treated as custom_ React components. By capitalizing the first letter of your component names, you ensure that React correctly identifies them as components and renders them accordingly.

- **JSX**: **_JSX (JavaScript XML)_** is a syntax extension for JavaScript that allows you to write **_HTML-like_** code **_within your JavaScript files_**. It is used in React to define the structure and appearance of components. JSX makes it easier to visualize the component's output and allows for a more intuitive way of building user interfaces.

### What is JSX?

#### Intro

- **JSX** is a syntax extension for JavaScript that allows you to write HTML-like code within your JavaScript files. It is used in React to define the structure and appearance of components. JSX makes it easier to visualize the component's output and allows for a more intuitive way of building user interfaces.

- _Why use JSX:_
    1. **Readability**: JSX allows you to write HTML-like code directly within your JavaScript files, making it easier to read and understand the structure of your components. This can improve the overall readability of your code and make it easier for developers to work with.

    2. **Declarative Syntax**: JSX uses a declarative syntax, which means you describe what you want to see on the screen, and React takes care of updating the UI to match that description. This makes it easier to understand and reason about your code, as you can focus on the desired outcome rather than the steps needed to achieve it.

    3. **Integration with JavaScript**: JSX allows you to seamlessly integrate JavaScript expressions and logic within your HTML-like code. You can use curly braces `{}` to embed JavaScript expressions, making it easy to dynamically render content based on the component's state or props.

    4. **Tooling and Ecosystem**: JSX is widely supported by modern development tools and has a rich ecosystem of libraries and plugins that enhance the development experience. This includes features like syntax highlighting, code formatting, and linting, which can help catch errors and enforce coding standards.

    5. **Performance**: JSX is compiled into efficient JavaScript code by tools like Babel, which optimizes the rendering process and improves performance. This allows developers to write expressive and maintainable code without sacrificing performance.

#### JSX Rules

1. **Return a single root element:** 
   - In JSX, a component must return a single root element. This means that all the elements you want to render must be wrapped in a single parent element, such as a `<div>`, `<section>`, or `<React.Fragment>`. If you try to return multiple sibling elements without a parent, you will encounter a syntax error.

   Example:
   ```jsx
   // Correct
   const MyComponent = () => {
       return (
           <div>
               <h1>Hello, World!</h1>
               <p>This is my component.</p>
           </div>
       );
   };

   // Incorrect
   const MyComponent = () => {
       return (
           <h1>Hello, World!</h1>
           <p>This is my component.</p>
       );
   };
   ```

2. **Close all tags:** 
   - In JSX, all HTML tags must be properly closed. This includes self-closing tags like `<img />`, `<input />`, and `<br />`. If you forget to close a tag, you will encounter a syntax error.

   Example:
   ```jsx
   // Correct
   const MyComponent = () => {
       return (
           <div>
               <img src="image.jpg" alt="My Image" />
               <input type="text" />
           </div>
       );
   };

   // Incorrect
   const MyComponent = () => {
       return (
           <div>
               <img src="image.jpg" alt="My Image">
               <input type="text">
           </div>
       );
   };
   ```

3. **camelCase most things:** 
   - In JSX, most attributes and event handlers use camelCase naming conventions. For example, instead of using `class` for CSS classes, you should use `className`, and instead of `onclick`, you should use `onClick`. This is because JSX is based on JavaScript, which uses camelCase for naming conventions.
   - JSX turns into JavaScript, and attributes of elements become keys of JavaScript objects, so you can’t use dashes or reserved words such as `class`. Because of this, many HTML attributes are written in camelCase. Instead of `stroke-width`, you’d use `strokeWidth`, and instead of `class` you’d use `className`.

   Example:
   ```jsx
   // Correct
   const MyComponent = () => {
       return (
           <div className="my-class" onClick={handleClick}>
               Click me!
           </div>
       );
   };

   // Incorrect
   const MyComponent = () => {
       return (
           <div class="my-class" onclick={handleClick}>
               Click me!
           </div>
       );
   };
   ```

### Passing Data Between components

#### Data Transfer in React - Props

- In React, **props** (short for **_properties_**) are a way to pass data from a parent component to a child component. Props allow you to customize the behavior and appearance of child components based on the data provided by the parent component. They are read-only, meaning that child components cannot modify the props they receive.

- Props are passed to child components as attributes in the JSX syntax. The parent component can define the props and their values, and the child component can access them through the `props` object.

- Props syntax:
    ```jsx
    // Parent Component
    const ParentComponent = () => {
        const message = "Hello from the parent component!";
        return (
            <ChildComponent message={message} />
        );
    };

    // Child Component
    const ChildComponent = (props) => {
        return (
            <div>
                <h1>{props.message}</h1>
            </div>
        );
    };
    ```
- In this example, the `ParentComponent` defines a prop called `message` and passes it to the `ChildComponent`. The `ChildComponent` receives the prop as an argument and can access its value using `props.message`. This allows the child component to display the message passed from the parent component.

- Another example: instead of defining 3 buttons with different text:
    ```jsx
    function Button(props) {
        const buttonStyle = {
            color: props.color,
            fontSize: props.fontSize + 'px'
        };

        return (
            <button style={buttonStyle}>{props.text}</button>
        );
        }

        export default function App() {
        return (
            <div>
            <Button text="Click Me!" color="blue" fontSize={12} />
            <Button text="Don't Click Me!" color="red" fontSize={12} />
            <Button text="Click Me!" color="blue" fontSize={20} />
            </div>
        );
    }
    ```
    - You can create a reusable `Button` component that accepts props for `text`, `color`, and `fontSize`. This allows you to easily customize the appearance and behavior of each button without having to duplicate code for each button instance.


#### Prop Destructuring 

- **Prop Desctructuring** is a technique in React that allows you to extract specific properties from the `props` object and assign them to variables for easier access within a component. This can make your code cleaner and more readable by reducing the need to repeatedly access properties through the `props` object.
- Example:
    ```jsx
    // Without Prop Destructuring
    const MyComponent = (props) => {
        return (
            <div>
                <h1>{props.title}</h1>
                <p>{props.description}</p>
            </div>
        );
    };

    // With Prop Destructuring
    const MyComponent = ({ title, description }) => {
        return (
            <div>
                <h1>{title}</h1>
                <p>{description}</p>
            </div>
        );
    };
    ```
    - In the first example, we access the `title` and `description` properties from the `props` object using `props.title` and `props.description`. In the second example, we use prop destructuring to extract the `title` and `description` properties directly in the function parameters, allowing us to access them without needing to reference the `props` object. This can make the code more concise and easier to read.

#### Default Props

- **Default Props** in React are a way to specify default values for props that are not provided by the parent component. This allows you to ensure that your components have valid data to work with, even if the parent component does not pass all the necessary props. Default props can be defined using the `defaultProps` property on a component.
- Example:
    ```jsx
    const MyComponent = ({ title, description }) => {
        return (
            <div>
                <h1>{title}</h1>
                <p>{description}</p>
            </div>
        );
    };

    MyComponent.defaultProps = {
        title: "Default Title",
        description: "Default Description"
    };

    export default MyComponent;
    ```
    - In this example, if the `MyComponent` is rendered without passing the `title` and `description` props, it will use the default values specified in `defaultProps`. This ensures that the component can still render properly even if the parent component does not provide all the necessary props.

#### Functions as Props

- Functions can also be passed as props in React, allowing you to pass behavior and logic from a parent component to a child component. This is commonly used for handling events or performing actions in response to user interactions. When a function is passed as a prop, the child component can call that function to trigger the desired behavior defined in the parent component.
- Example:
    ```jsx
    // Parent Component
    const ParentComponent = () => {
        const handleClick = () => {
            alert("Button clicked!");
        };

        return (
            <ChildComponent onButtonClick={handleClick} />
        );
    };

    // Child Component
    const ChildComponent = ({ onButtonClick }) => {
        return (
            <button onClick={onButtonClick}>Click Me!</button>
        );
    };
    ```
    - In this example, the `ParentComponent` defines a function called `handleClick` that displays an alert when called. This function is passed to the `ChildComponent` as a prop named `onButtonClick`. The `ChildComponent` receives the function as a prop and assigns it to the `onClick` event handler of a button. When the button is clicked, it triggers the `handleClick` function defined in the parent component, demonstrating how functions can be passed as props to enable communication and interaction between components in React.

### Keys in React

- In React, **Keys** are special attributes that you need to include when creating lists of elements. They help React identify which items have changed, been added, or removed, allowing it to efficiently update the UI when the list changes. Keys should be unique among siblings and are typically assigned using a unique identifier from your data, such as an ID or index.
- Example:
    ```jsx
    const MyList = ({ items }) => {
        return (
            <ul>
                {items.map((item) => (
                    <li key={item.id}>{item.name}</li>
                ))}
            </ul>
        );
    };
    ```
    - In this example, we have a `MyList` component that takes an array of `items` as a prop. When rendering the list of items, we use the `map` function to create a list item (`<li>`) for each item in the array. We assign a unique `key` prop to each list item using `item.id`, which helps React efficiently update the list when items are added, removed, or changed. This is important for performance and ensures that React can correctly identify and manage the elements in the list.

- Key syntax:
    ```jsx
    <Component key={keyValue}/>
    <div key={keyValue}>...</div>
    ```
